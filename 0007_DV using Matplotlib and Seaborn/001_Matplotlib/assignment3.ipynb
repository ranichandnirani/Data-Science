**Assignment:**  
Create a pie chart showing the market share of mobile OS (Android, iOS, others). 
    
Then recreate the same using a horizontal bar chart and observe which is easier to understand.


```python
# Pie chart
import matplotlib.pyplot as plt

# Data
os = ["Android", "iOS", "Others"]
market_share = [70, 25, 5]

plt.pie(market_share, labels=os, autopct='%1.1f%%', startangle=100)

plt.title("Mobile OS Market Share")
plt.show()
```

```python
# 2. Horizontal Bar Chart

plt.figure(figsize=(8, 5))

plt.barh(os, market_share)

plt.xlabel("Market Share (%)")
plt.ylabel("Mobile OS")
plt.title("Mobile OS Market Share")

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
```
